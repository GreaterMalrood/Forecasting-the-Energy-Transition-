#**PRODUCTION**

```python
sns.set_theme(style="darkgrid")

plt.figure(figsize=(15, 8))

# Convert 'Year' to numeric type for proper plotting
prodmelted['Year'] = pd.to_numeric(prodmelted['Year'])

sns.lineplot(data=prodmelted, x='Year', y='Value', hue='MSN', marker='o')

plt.title('Production Value by MSN Over Years', fontsize=15)
plt.xlabel('Year')
plt.ylabel('Production Value')
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left', borderaxespad=0.) # Adjust legend position to avoid overlap
plt.tight_layout() # Adjust layout to prevent labels from being cut off
plt.show()
```

```python
sns.set_theme(style="darkgrid")

# Create a FacetGrid
# col="MSN" creates a new graph for each unique MSN
# col_wrap=3 limits the number of graphs per row
# sharey=False allows each graph to have its own y-axis scale if values differ greatly
g = sns.FacetGrid(prodmelted, col="MSN", col_wrap=3, height=4, aspect=1.2, sharey=False)

# Map the lineplot to the grid
g.map_dataframe(sns.lineplot, x='Year', y='Value', marker='o')

# Add titles and labels
g.set_axis_labels("Year", "Production Value")
g.set_titles(col_template="{col_name}")

# Adjust layout
g.tight_layout()
plt.show()
```
